```python
import csv
import json
import pandas as pd
```

```python
# Working with messy data
data = [
    {"student_id": 1, "major": "Art History", "GPA": 3, "is_cs_major": "No", "credits_taken": '10.5'},
    {"student_id": 2, "major": "Computer Science", "GPA": 3.3, "is_cs_major": "Yes", "credits_taken": '13'},
    {"student_id": 3, "major": "Math", "GPA": 4.0, "is_cs_major": "No", "credits_taken": '14'},
    {"student_id": 4, "major": "Computer Science", "GPA": 2.5, "is_cs_major": "Yes", "credits_taken": '9.5'},
    {"student_id": 5, "major": "English", "GPA": 2, "is_cs_major": "No", "credits_taken": '17'},

]
with open("raw_survey_data.csv", mode="w", newline="") as file:
   writer = csv.DictWriter(file, fieldnames=["student_id", "major", "GPA", "is_cs_major", "credits_taken"])
   writer.writeheader()
   writer.writerows(data)
```

```python
# Working with JSON data
courses = [
  {
    "course_id": "DS2002",
    "section": "001",
    "title": "Data Science Systems",
    "level": 200,
    "instructors": [
      {"name": "Austin Rivera", "role": "Primary"}, 
      {"name": "Heywood Williams-Tracy", "role": "TA"} 
    ]
  },
  {
    "course_id": "DS3100",
    "title": "Foundations of Machine Learning",
    "section": "002",
    "level": 300,
    "instructors": [
      {"name": "Michael Freenor", "role": "Primary"}
    ]
  },
]

with open("raw_course_catalog.json", mode="w") as f:
    json.dump(courses, f)
```

```python
# Cleaning and Validating Data
df = pd.read_csv("raw_survey_data.csv")
df["is_cs_major"] = df["is_cs_major"].replace({"Yes": True, "No": False}).astype(bool)
df = df.astype({"GPA": "float64", "credits_taken": "float64"})
df.to_csv("cleaned_survey_data.csv", index=False)
```

```python
# Normalize to JSON 
with open("raw_course_catalog.json","r") as f:
    json_data = json.load(f)

classes = pd.json_normalize(
    json_data, 
    record_path=["instructors"],
    meta=["course_id", "title", "level", "section"], 
)
classes.to_csv("clean_course_catalog.csv", index=False)
```
